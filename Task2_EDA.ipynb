```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Fetch live data
url = "http://books.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles, prices, availabilities = [], [], []
books = soup.find_all("article", class_="product_pod")

for book in books:
    titles.append(book.h3.a["title"])
    prices.append(book.find("p", class_="price_color").text)
    availabilities.append(book.find("p", class_="instock availability").text.strip())

# Create DataFrame
df = pd.DataFrame({
    "Book Title": titles,
    "Price": prices,
    "Availability": availabilities
})

print("--- 1. FIRST LOOK AT DATA ---")
print(df.head())
print("\n" + "="*40 + "\n")

print("--- 2. DATASET INFO ---")
print(df.info())
print("\n" + "="*40 + "\n")

print("--- 3. MISSING VALUES CHECK ---")
print(df.isnull().sum())
print("\n" + "="*40 + "\n")

# 4. Data Cleaning: Clean Price Column (Convert string currency to float number)
df["Clean_Price"] = df["Price"].str.replace("Â", "").str.replace("£", "").astype(float)

print("--- 4. STATISTICAL SUMMARY OF PRICES ---")
print(df["Clean_Price"].describe())

# Save cleaned CSV
df.to_csv("cleaned_books.csv", index=False)
print("\n✅ Cleaned dataset saved as 'cleaned_books.csv'!")
```
