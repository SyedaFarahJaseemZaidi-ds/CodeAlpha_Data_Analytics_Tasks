```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Define target URL
url = "http://books.toscrape.com/"

# 2. Fetch page HTML
response = requests.get(url)

# 3. Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# 4. Extract Data
titles = []
prices = []
availabilities = []

books = soup.find_all("article", class_="product_pod")

for book in books:
    titles.append(book.h3.a["title"])
    prices.append(book.find("p", class_="price_color").text)
    availabilities.append(book.find("p", class_="instock availability").text.strip())

# 5. Create DataFrame and Save CSV
df = pd.DataFrame({
    "Book Title": titles,
    "Price": prices,
    "Availability": availabilities
})

df.to_csv("scraped_books.csv", index=False)
print("Scraping complete! Data saved to scraped_books.csv.")
```
